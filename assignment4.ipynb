```python
import numpy as np
import pandas as pd
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
data = [
    [1, 0.20, 0.10, 0.80, 0.10, 1],
    [2, 0.30, 0.20, 0.70, 0.20, 1],
    [3, 0.40, 0.15, 0.90, 0.15, 1],
    [4, 0.50, 0.30, 0.80, 0.10, 1],
    [5, 0.60, 0.25, 0.75, 0.20, 1],
    [6, 0.70, 0.35, 0.85, 0.10, 1],
    [7, 0.80, 0.40, 0.90, 0.15, 1],
    [8, 0.25, 0.50, 0.70, 0.40, 1],
    [9, 0.35, 0.45, 0.80, 0.35, 1],
    [10, 0.45, 0.55, 0.75, 0.45, 1],
    [11, 0.55, 0.50, 0.85, 0.30, 1],
    [12, 0.65, 0.60, 0.80, 0.40, 1],
    [13, 0.20, 0.80, 0.40, 0.75, 0],
    [14, 0.30, 0.70, 0.35, 0.80, 0],
    [15, 0.40, 0.85, 0.45, 0.70, 0],
    [16, 0.50, 0.75, 0.30, 0.85, 0],
    [17, 0.60, 0.90, 0.40, 0.80, 0],
    [18, 0.70, 0.80, 0.35, 0.90, 0],
    [19, 0.80, 0.85, 0.45, 0.75, 0],
    [20, 0.90, 0.70, 0.30, 0.85, 0],
    [21, 0.25, 0.90, 0.50, 0.80, 0],
    [22, 0.45, 0.95, 0.40, 0.90, 0],
    [23, 0.65, 0.85, 0.30, 0.95, 0],
    [24, 0.85, 0.90, 0.35, 0.85, 0]
]
columns = [
    "ID", "Vibration", "Slip",
    "SurfaceGrip", "ObstacleRate", "Stable"
]
df = pd.DataFrame(data, columns=columns)
print("A. Деректер жиынтығы")
print(df.to_string(index=False))
```

```python
# 2. Оқыту және тест деректері
train = df[df["ID"] <= 18]
test = df[df["ID"] >= 19]

features = [
    "Vibration", "Slip",
    "SurfaceGrip", "ObstacleRate"
]

X_train = train[features]
X_test = test[features]
y_train = train["Stable"]
y_test = test["Stable"]

print("\nОқыту:", len(train))
print("Тест:", len(test))
print("Кластар:", train["Stable"].value_counts().to_dict())
```

```python
# 3. Decision Tree
tree = DecisionTreeClassifier(random_state=42)
tree.fit(X_train, y_train)
tree_pred = tree.predict(X_test)

def show_metrics(name, predictions):
    print("\n", name)
    print("Accuracy:", accuracy_score(y_test, predictions))
    print("Precision:", precision_score(
        y_test, predictions, pos_label=0, zero_division=0
    ))
    print("Recall:", recall_score(
        y_test, predictions, pos_label=0, zero_division=0
    ))
    print("F1:", f1_score(
        y_test, predictions, pos_label=0, zero_division=0
    ))
    print("Confusion Matrix:")
    print(confusion_matrix(y_test, predictions, labels=[0, 1]))

show_metrics("B. Decision Tree", tree_pred)
```

```python
# 4. Алдын ала өңдеу
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# 5. Нейрондық желі - 3 нейрон
mlp3 = MLPClassifier(
    hidden_layer_sizes=(3,),
    solver="lbfgs",
    random_state=42,
    max_iter=5000
)

mlp3.fit(X_train_scaled, y_train)
pred3 = mlp3.predict(X_test_scaled)

show_metrics("C. MLP - 3 нейрон", pred3)
```

```python
# 6. Нейрондық желі - 6 нейрон
mlp6 = MLPClassifier(
    hidden_layer_sizes=(6,),
    solver="lbfgs",
    random_state=42,
    max_iter=5000
)

mlp6.fit(X_train_scaled, y_train)
pred6 = mlp6.predict(X_test_scaled)

show_metrics("C. MLP - 6 нейрон", pred6)
```

```python
# 7. Эксперименттерді салыстыру
comparison = pd.DataFrame({
    "Нейрон саны": [3, 6],
    "Accuracy": [
        accuracy_score(y_test, pred3),
        accuracy_score(y_test, pred6)
    ],
    "F1": [
        f1_score(y_test, pred3, pos_label=0),
        f1_score(y_test, pred6, pos_label=0)
    ]
})

print("\nC. Эксперимент нәтижелері")
print(comparison.to_string(index=False))
```

```python
# 8. Барлық тест нәтижелері
results = pd.DataFrame({
    "Test ID": test["ID"].values,
    "Нақты класс": y_test.values,
    "Decision Tree": tree_pred,
    "MLP (3)": pred3,
    "MLP (6)": pred6
})

results["Дұрыс па?"] = np.where(
    (results["Decision Tree"] == results["Нақты класс"]) &
    (results["MLP (3)"] == results["Нақты класс"]) &
    (results["MLP (6)"] == results["Нақты класс"]),
    "Иә",
    "Жоқ"
)

print("\nD. Тест нәтижелері")
print(results.to_string(index=False))
```
